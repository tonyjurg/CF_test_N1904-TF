## Load the code

```python
from cfabric.core.fabric import Fabric
```

```python
# Load the N1904-TF dataset
CF = Fabric(locations='centerblc/N1904/tf/1.0.0')
```

```python
# Explore available features
CF.explore()
```

```python
# Load commonly used features
api = CF.loadAll()
```

```python
# Set up convenient references
F = api.F  # Node features
E = api.E  # Edge features
T = api.T  # Text API
L = api.L  # Locality API
N = api.N  # Node navigation
```

## Surface text of John 1:1

```python
# Get John 1:1
john_1_1 = T.nodeFromSection(('John', 1, 1))
print(f"Verse node: {john_1_1}")
print(f"\nGreek text: {T.text(john_1_1)}")
```

```python
# get the number of the verse node
john_1_1
```

## Grammatical features of John 1:1

```python
# Show each word with grammatical features
words = L.d(john_1_1, otype='word')
print("Word-by-word analysis of John 1:1:\n")
print(f"{'greek':<15} {'Lexeme':<12} {'Gloss':<30} {'POS':<8} {'Gender':<8} {'Number'}")
print("-" * 80)
for w in words:
    greek = F.text.v(w)
    lexeme = F.lemma.v(w) or F.lemma.v(w)
    gloss = F.gloss.v(w) or ''
    pos = F.sp.v(w) or ''
    gender = F.gender.v(w) or ''
    number = F.number.v(w) or ''
    print(f"{greek:<15} {lexeme:<12} {gloss:<30} {pos:<8} {gender:<8} {number}")
```

## Clause Structure of John 1:1

```python
# Explore clause and phrase structure of Genesis 1:1
clauses = L.d(john_1_1, otype='clause')
print(f"Clauses in John 1:1: {len(clauses)}\n")

for clause in clauses:
    clause_type = F.typ.v(clause)
    print(f"Clause (type={clause_type}):")
    
    phrases = L.d(clause, otype='phrase')
    for phrase in phrases:
        phrase_type = F.typ.v(phrase)
        phrase_func = F.function.v(phrase)
        phrase_words = [F.text.v(w) for w in L.d(phrase, otype='word')]
        print(f"  Phrase [{phrase_func}] ({phrase_type}): {' '.join(phrase_words)}")
```

## Some Corpus Statistics

```python
# Corpus statistics
from collections import Counter

# Count node types
otype_counts = Counter()
for n in N.walk():
    otype_counts[F.otype.v(n)] += 1

print("N1904 Corpus Statistics:\n")
print(f"{'Node Type':<20} {'Count':>10}")
print("-" * 32)
for otype, count in sorted(otype_counts.items(), key=lambda x: -x[1]):
    print(f"{otype:<20} {count:>10,}")
```
